```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sentences=['the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good']
```

```python
sentences
```

```python
vocab_size=10000
```

```python
onehot_encoder=[]
for sent in sentences:
    encodr=one_hot(sent,10000)
    onehot_encoder.append(encodr)
onehot_encoder
```

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
sent_length=(len(max(onehot_encoder))+4)
embedded_docs=pad_sequences(onehot_encoder,padding="pre",maxlen=sent_length)
print(embedded_docs)
```

```python
## Feature representation
dim=10
```

```python
model=Sequential()
model.add(Embedding(vocab_size,dim,input_length=sent_length))
model.compile("adam","mse")
```

```python
model.summary()
```

```python
model.predict(embedded_docs[0])
```
